```python
import random 
import numpy as np 
import scipy
import math
import matplotlib.pyplot as plt

N=10000
n=10000

X_n = 2*np.random.default_rng().binomial(n, 0.5, size=N) - n
    
print(np.var(X_n))

scipy.stats.kstest(X_n/math.sqrt(n),"norm")
plt.hist(X_n/math.sqrt(n), 50, density=True)
```

```python
from scipy.stats import norm

Z = X_n / math.sqrt(n)           # X_n standardisé
x = np.linspace(-4, 4, 200)      # grille pour la densité

plt.hist(Z, bins=50, density=True, alpha=0.6, label="X_n / √n (simulé)")
plt.plot(x, norm.pdf(x), linewidth=2, label="Densité N(0, 1)")
plt.xlabel("valeur")
plt.ylabel("densité")
plt.title("Marche aléatoire normalisée vs loi normale")
plt.legend()
plt.show()

print(scipy.stats.kstest(Z, "norm"))
```

```python
T = 1
n = 1000
m = 10
dt = T/n

rng = np.random.default_rng(42)

Z = rng.standard_normal((n, m))
dW = Z * np.sqrt(dt)
W = np.cumsum(dW, axis = 0)
A = np.zeros((1,m))
W_t = np.concatenate((A,W))

time = np.linspace(0,T,n+1)

plt.plot(time,W_t)

```
