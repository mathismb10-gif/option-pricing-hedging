```python
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.path.append("..")
from src.payoffs import call_payoff, put_payoff

S_T = np.linspace(50, 150, 201)

print(call_payoff(S_T, 100))
print(call_payoff(S_T, 100).shape)
```

```python
np.allclose(call_payoff(S_T,100)-put_payoff(S_T,100),S_T - 100)
```

```python
plt.plot(S_T,call_payoff(S_T,100), label="Call payoff (K=100)")
plt.plot(S_T,put_payoff(S_T,100), label="Put payoff (K=100)")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Underlying price at maturity S_T")
plt.ylabel("Payoff")
plt.title("Payoffs at maturity")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.plot(S_T,call_payoff(S_T,100)-5, label="Long call P&L (premium 5)")
plt.plot(S_T,put_payoff(S_T,100)-4, label="Long put P&L (premium 4)")
plt.plot(S_T,-call_payoff(S_T,100)+5, label="Short call P&L (premium 5)")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Underlying price at maturity S_T")
plt.ylabel("P&L")
plt.title("P&L at maturity")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
