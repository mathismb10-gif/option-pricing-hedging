```python
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.path.append("..")
from src.payoffs import call_payoff, put_payoff

S_T = np.linspace(50, 150, 201)

print(call_payoff(S_T, 100))
print(call_payoff(S_T, 100).shape)
```

```python
np.allclose(call_payoff(S_T,100)-put_payoff(S_T,100),S_T - 100)
```

```python
plt.plot(S_T,call_payoff(S_T,100), label="Payoff call (K=100)")
plt.plot(S_T,put_payoff(S_T,100), label="Payoff put (K=100)")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Prix du sous-jacent à maturité S_T")
plt.ylabel("Payoff (€)")
plt.title("Payoffs à maturité")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.plot(S_T,call_payoff(S_T,100)-5, label="P&L call long (prime 5)")
plt.plot(S_T,put_payoff(S_T,100)-4, label="P&L put long (prime 4)")
plt.plot(S_T,-call_payoff(S_T,100)+5, label="P&L call short (prime 5)")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Prix du sous-jacent à maturité S_T")
plt.ylabel("P&L (€)")
plt.title("P&L à maturité")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
